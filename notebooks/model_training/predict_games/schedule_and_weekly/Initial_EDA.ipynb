```python
from src.data.collect_all import get_schedule_and_weekly_data
schedule_and_weekly_df = get_schedule_and_weekly_data([2019, 2021, 2022])
schedule_and_weekly_df.head()
```

```python
columns = [
    'game_id',
    'season',
    'game_type',
    'week',
    'opp_team',
    'opp_score',
    'target_team',
    'target_score',
    'div_game',
    'target_days_since_previous_game',
    # 'bye_week_target',
    # 'bye_week_opp',
    # 'target_days_to_bye_week',
    # 'opp_days_to_bye_week',
    'target_game_count',
    'opp_game_count',
    'is_home_target',
    'target_win'
]
schedule_and_weekly_df[schedule_and_weekly_df['target_team'] == 'CAR'][columns].sort_values('week')
```

```python
# Find the missing week for each team for each year
# Ensure you're only using regular season games
regular_season_df = schedule_and_weekly_df[schedule_and_weekly_df['game_type'] == 'REG']

# Get unique seasons and teams
seasons = regular_season_df['season'].unique()
teams = pd.unique(regular_season_df[['target_team', 'opp_team']].values.ravel())

bye_weeks = []

for season in seasons:
    season_df = regular_season_df[regular_season_df['season'] == season]
    weeks = season_df['week'].unique()

    for team in teams:
        for week in weeks:
            week_df = season_df[season_df['week'] == week]
            # Check if team plays this week (either as target or opp)
            plays_this_week = (
                (week_df['target_team'] == team) |
                (week_df['opp_team'] == team)
            ).any()

            if not plays_this_week:
                bye_weeks.append({
                    'team': team,
                    'season': season,
                    'bye_week': week
                })
                break  # Found the bye week, no need to check more weeks for this team

# Convert to DataFrame
bye_week_df = pd.DataFrame(bye_weeks)
bye_week_df
```
